```python
# Import pandas for dataframe operations
import pandas as pd

# Import numpy for numerical operations
import numpy as np

```

```python
#Task 1
There are three different tables as given above. Please make three dataframe in python
and save them as three .csv files. From Task 2 to Task 10, use the saved .csv files only.
```

```python
# Employee DataFrame 

# Creating employee data as dictionary
employee_data = {
    "ID": ["A001", "A002", "A003", "A004", "A005"],   # Employee IDs
    "Name": ["John Alter", "Alice Luxumberg", "Tom Sabestine", "Nina Adgra", "Amy Johny"],  # Names
    "Gender": ["M", "F", "M", "F", "F"],              # Gender
    "City": ["Paris", "London", "Berlin", "Newyork", "Madrid"],  # Cities
    "Age": [25, 27, 29, 31, 30]                       # Age
}

# Converting dictionary to DataFrame
employee_df = pd.DataFrame(employee_data)

# Saving employee dataframe to CSV
employee_df.to_csv("Employee.csv", index=False)
```

```python
#  Seniority DataFrame 

# Creating seniority level data
seniority_data = {
    "ID": ["A001", "A002", "A003", "A004", "A005"],   # Employee IDs
    "Designation_Level": [2, 2, 3, 2, 3]              # Designation levels
}

# Converting dictionary to DataFrame
seniority_df = pd.DataFrame(seniority_data)

# Saving seniority dataframe to CSV
seniority_df.to_csv("Seniority.csv", index=False)
```

```python
#  Project DataFrame

# Creating project data (None represents missing cost)
project_data = {
    "ID": ["A001","A002","A003","A004","A005","A002","A005","A003","A001","A003","A001","A004","A004","A005"],
    "Project": ["Project 1","Project 2","Project 3","Project 4","Project 5","Project 6",
                "Project 7","Project 8","Project 9","Project 10","Project 11","Project 12","Project 13","Project 14"],
    "Project_Cost": [1002000,2000000,4500000,5500000,None,680000,400000,350000,None,300000,2000000,1000000,3000000,200000],
    "Status": ["Finished","Ongoing","Finished","Ongoing","Finished","Failed",
               "Finished","Failed","Ongoing","Finished","Failed","Ongoing","Finished","Finished"]
}

# Converting dictionary to DataFrame
project_df = pd.DataFrame(project_data)

# Saving project dataframe to CSV
project_df.to_csv("Project.csv", index=False)
```

```python
#Task 2
The cost column in the dataframe “Project” has some missing values. Your task is to
compute these missing values. Replace the missing values by running average. You
should use the “For” loop for this task.
```

```python
# Reading Project CSV file
project_df = pd.read_csv("Project.csv")

# Initializing sum and count for running average
total = 0
count = 0

# Looping through Project_Cost column
for i in range(len(project_df)):
    
    # Checking if value is not missing
    if not pd.isna(project_df.loc[i, "Project_Cost"]):
        total += project_df.loc[i, "Project_Cost"]  # Add cost
        count += 1                                  # Increase count
    else:
        # Replace missing value with running average
        project_df.loc[i, "Project_Cost"] = total / count

# Display updated project dataframe
project_df
```

```python
#Task 3
Split the name column in the Employee dataframe into two new columns “First Name”,
and “LastName” and remove the older “name” column.
```

```python
# Reading Employee CSV
employee_df = pd.read_csv("Employee.csv")

# Splitting Name column into First and Last name
employee_df[["First_Name", "Last_Name"]] = employee_df["Name"].str.split(" ", expand=True)

# Dropping old Name column
employee_df.drop("Name", axis=1, inplace=True)

# Display updated employee dataframe
employee_df
```

```python
#Task 4
Join all three dataframes in one single dataframe. Name it “Final”
```

```python
# Reading Seniority CSV
seniority_df = pd.read_csv("Seniority.csv")

# Merging Employee and Seniority dataframes
final_df = pd.merge(employee_df, seniority_df, on="ID")

# Merging result with Project dataframe
final_df = pd.merge(final_df, project_df, on="ID")

# Display Final dataframe
final_df
```

```python
#Task 5
Add a new bonus column in the Final dataframe. Give a 5% bonus concerning project
cost only to employees who have finished the projects.
```

```python
# Creating Bonus column with 5% bonus for Finished projects
final_df["Bonus"] = np.where(
    final_df["Status"] == "Finished",
    final_df["Project_Cost"] * 0.05,
    0
)

# Display Final dataframe
final_df
```

```python
#Task 6
Demote the designation level by 1, whose projects have status “fail”. Delete the
employees record whose designation level is above 4.
```

```python
# Demoting designation level by 1 if project failed
final_df.loc[final_df["Status"] == "Failed", "Designation_Level"] -= 1

# Removing employees whose designation level is above 4
final_df = final_df[final_df["Designation_Level"] <= 4]

# Display Final dataframe
final_df
```

```python
#Task 7
Add “Mr.” and “Mrs.” to the first name column and drop the gender column.
```

```python
# Adding title based on gender
final_df["First_Name"] = np.where(
    final_df["Gender"] == "M",
    "Mr. " + final_df["First_Name"],
    "Mrs. " + final_df["First_Name"]
)

# Dropping Gender column
final_df.drop("Gender", axis=1, inplace=True)

# Display updated dataframe
final_df
```

```python
#Task 8
Promote designation level by 1 for the employees whose age is more than 29 years using
IF condition.
```

```python
# Promoting employees whose age is greater than 29
final_df.loc[final_df["Age"] > 29, "Designation_Level"] += 1

# Display updated dataframe
final_df
```

```python
#Task 9
Add the cost of all projects for each Employee and save it in new dataframe
“TotalProjCost” with three columns ID, First Name, and Total cost
```

```python
# Grouping by ID and First Name to calculate total project cost
total_proj_cost = final_df.groupby(
    ["ID", "First_Name"]
)["Project_Cost"].sum().reset_index()

# Renaming column
total_proj_cost.rename(columns={"Project_Cost": "Total_Project_Cost"}, inplace=True)

# Display Total Project Cost dataframe
total_proj_cost
```
